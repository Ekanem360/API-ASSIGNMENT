```python
import requests
```

```python
res1 = requests.get("https://api.frankfurter.dev/v1/latest?from=USD").json()
base = res1.get("base")
date = res1.get("date")
total_currencies = len(res1.get("rates", {}))

print(f"Task 1 Results:")
print(f"Base: {base} | Date: {date} | Currencies Returned: {total_currencies}\n")

```

```python
res2 = requests.get("https://api.frankfurter.dev/v1/currencies").json()
print("Task 2 Results:")
print(f"Total Supported Currencies: {len(res2)}")
print(f"Is NGN supported? {'NGN' in res2}")
print("Supported Currencies:")
for code, name in res2.items():
    print(f" {code}: {name}")
print("\n")
```

```python
def get_rates(base_currency):
    try:
        url = f"https://api.frankfurter.dev/v1/latest?from={base_currency}"
        response = requests.get(url, timeout=10)
        
        if response.status_code == 200:
            return response.json()
        else:
            # Handle API error response gracefully
            error_data = response.json() if response.headers.get("content-type") == "application/json" else {}
            msg = error_data.get("message", "Not found or invalid currency")
            return f"Failed (Status {response.status_code}): {msg}"
    except Exception as e:
        return f"Request failed: {str(e)}"

currencies_to_test = ["USD", "EUR", "GBP", "JPY", "CHF", "NGN", "ZWL", "XYZ", "ng"]

print("Task 3 Results:")
for code in currencies_to_test:
    print(f"{code}: {get_rates(code)}")
print("\n")

```

```python
users_data = requests.get("https://jsonplaceholder.typicode.com/users").json()

users_dict = {
    user["id"]: {
        "name": user["name"],
        "email": user["email"],
        "city": user["address"]["city"]
    }
    for user in users_data
}

print("Task 4 Results:")
print(users_dict)
print("\n")
```

```python
post_payload = {"title": "New Post", "body": "Sample content", "userId": 1}
post_res = requests.post("https://jsonplaceholder.typicode.com/posts", json=post_payload)

status_code = post_res.status_code
created_id = post_res.json().get("id")

get_res = requests.get(f"https://jsonplaceholder.typicode.com/posts/{created_id}")

print("Task 5 Results:")
print(f"POST Status Code: {status_code} | Generated ID: {created_id}")
print(f"GET Status Code for ID {created_id}: {get_res.status_code}")
```

For number 5: Fetching post id 101 returns HTTP 404 Not Found response. This happens because Jsonplaceholder is a mock rest api that fakes creation responses without persisting any new resources to a server database
